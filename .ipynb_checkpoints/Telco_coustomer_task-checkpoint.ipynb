```python
#### Total Charges Column Is Object DataTYpe -- so Convert it into Numeric 

df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce').fillna(0)
```

```python
df['Total Charges'].unique()
```

```python
df.info()
```

```python
# 1. Drop columns that are IDs, single-value, or cause target leakage
cols_to_drop = [
    'CustomerID', 'Count', 'Country', 'State', 'City', 'Zip Code', 
    'Lat Long', 'Latitude', 'Longitude', 'Churn Label', 
    'Churn Score', 'CLTV', 'Churn Reason'
]
df = df.drop(columns=cols_to_drop)
```

```python

# 2. Binary mapping (2-category columns -> 0/1)

binary_cols = ['Gender', 'Senior Citizen', 'Partner', 'Dependents', 'Phone Service', 'Paperless Billing']

df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
for col in ['Senior Citizen', 'Partner', 'Dependents', 'Phone Service', 'Paperless Billing']:
    df[col] = df[col].map({'Yes': 1, 'No': 0})
```

```python
# Automatically find and encode all remaining text/categorical columns
cat_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()

if cat_cols:
    df = pd.get_dummies(df, columns=cat_cols, drop_first=True, dtype=int)

df.info()
```

```python

# Verify all columns are numeric
df.info()
```

#### •	Detect and handle duplicate records.

```python
df.duplicated().sum()
```

```python
# Re-fetch the original values for the broken binary columns from the excel sheet
df_raw = pd.read_excel('Telco_customer_churn.xlsx')

binary_cols = ['Gender', 'Senior Citizen', 'Partner', 'Dependents', 'Phone Service', 'Paperless Billing']

for col in binary_cols:
    df[col] = df_raw.loc[df.index, col].astype(str).str.strip().map({
        'Male': 1, 'Female': 0,
        'Yes': 1, 'No': 0
    })

df[binary_cols].info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```
