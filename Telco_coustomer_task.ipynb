### •	Detect outliers.

```python
# 1. Visualize continuous numerical features using boxplots
plt.figure(figsize=(15, 4))
for i, col in enumerate(num_features, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=X[col], color='skyblue')
    plt.title(f'Boxplot of {col}')
plt.tight_layout()
plt.show()

# 2. IQR Method to count outliers per feature
print("Outlier counts using IQR method (1.5 * IQR):")
for col in num_features:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = X[(X[col] < lower_bound) | (X[col] > upper_bound)]
    print(f"- {col}: {len(outliers)} outliers")
```

#### Detect The Outlier Using LOF 

```python
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
x_scaled = ss.fit_transform(X)
```

```python
lof = LocalOutlierFactor(
    n_neighbors = 20 ,
    contamination =0.05 
)
```

```python
lof
```

```python
lof_labels = pd.Series(lof.fit_predict(x_scaled))
lof_labels.isna().sum()
```

```python
X['lof_labels'] = lof_labels
```

```python
X['lof_labels'].value_counts()
```

```python
X['lof_labels'].value_counts().plot(kind ='bar')
```

```python
X['lof_labels'] = lof.fit_predict(x_scaled)

new_data = X[X['lof_labels'] == 1].copy()

y_new = y.loc[new_data.index]


new_data.drop(columns=['lof_labels'], inplace=True)

print(f"Original shape: {X.drop(columns=['lof_labels']).shape}")
print(f"New shape after removing outliers: {new_data.shape}")
print(f"New target shape: {y_new.shape}")
```

```python
y.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(new_data, y_new, test_size=0.2, random_state=42)

smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

print(f"Balanced Training Target Counts:\n{y_train_balanced.value_counts()}")
```

```python
# 1. Fit and transform the balanced training data
X_train_scaled = ss.fit_transform(X_train_balanced)

# 2. Transform the test data (using the mean/std from the training set)
X_test_scaled = ss.transform(X_test)
```

#### Using KNN Model Accuracy 

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train_scaled, y_train_balanced)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
```

#### Cross Validation 

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.svm import SVC

# 1. Initialize 5-fold stratified cross-validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Base SVM model
svm_cv = SVC(kernel='rbf', C=1.0, random_state=42)

# 3. Calculate cross-validation scores
cv_accuracy = cross_val_score(svm_cv, X_train_scaled, y_train_balanced, cv=skf, scoring='accuracy', n_jobs=-1)
cv_roc_auc = cross_val_score(svm_cv, X_train_scaled, y_train_balanced, cv=skf, scoring='roc_auc', n_jobs=-1)

# 4. Display results per fold and mean performance
print("=== 5-Fold Cross-Validation Results (SVM) ===")
print("Accuracy across folds:", [round(score, 4) for score in cv_accuracy])
print(f"Mean Accuracy: {cv_accuracy.mean():.4f} (+/- {cv_accuracy.std():.4f})")
print(f"Mean ROC-AUC:  {cv_roc_auc.mean():.4f} (+/- {cv_roc_auc.std():.4f})")
```

#### Hyperparameter Tuning 

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# 1. Define hyperparameter grid for SVM
param_grid = {
    'C': [0.1, 1, 5, 10],
    'gamma': ['scale', 'auto', 0.01, 0.1],
    'kernel': ['rbf']
}

# 2. Set up GridSearchCV with 5-fold cross-validation
svm_grid = GridSearchCV(
    estimator=SVC(probability=True, random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

# 3. Fit GridSearch on the scaled & balanced training set
svm_grid.fit(X_train_scaled, y_train_balanced)

# 4. Display best parameters and validation score
print("=== Hyperparameter Tuning Results ===")
print("Best Parameters:", svm_grid.best_params_)
print(f"Best CV ROC-AUC Score: {svm_grid.best_score_:.4f}")
```

#### Compare Before and After Tuning 

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 1. Best model from GridSearchCV
best_svm = svm_grid.best_estimator_

# 2. Predictions after tuning
y_pred_tuned = best_svm.predict(X_test_scaled)
y_pred_proba_tuned = best_svm.predict_proba(X_test_scaled)[:, 1]

# 3. Create comparison DataFrame
comparison_df = pd.DataFrame({
    'Metric': ['Accuracy', 'ROC-AUC', 'Precision', 'Recall', 'F1-Score'],
    'Before Tuning (Default SVM)': [
        accuracy_score(y_test, y_pred_svm),
        roc_auc_score(y_test, y_pred_proba_svm),
        precision_score(y_test, y_pred_svm),
        recall_score(y_test, y_pred_svm),
        f1_score(y_test, y_pred_svm)
    ],
    'After Tuning (Best SVM)': [
        accuracy_score(y_test, y_pred_tuned),
        roc_auc_score(y_test, y_pred_proba_tuned),
        precision_score(y_test, y_pred_tuned),
        recall_score(y_test, y_pred_tuned),
        f1_score(y_test, y_pred_tuned)
    ]
})

# Round values for display
comparison_df['Before Tuning (Default SVM)'] = comparison_df['Before Tuning (Default SVM)'].round(4)
comparison_df['After Tuning (Best SVM)'] = comparison_df['After Tuning (Best SVM)'].round(4)
comparison_df['Improvement'] = (comparison_df['After Tuning (Best SVM)'] - comparison_df['Before Tuning (Default SVM)']).round(4)

comparison_df
```

```python
# 1. Predictions on both Training and Testing sets
train_pred_tuned = best_svm.predict(X_train_scaled)
test_pred_tuned = best_svm.predict(X_test_scaled)

train_proba_tuned = best_svm.predict_proba(X_train_scaled)[:, 1]
test_proba_tuned = best_svm.predict_proba(X_test_scaled)[:, 1]

# 2. Performance metrics comparison
overfit_check = pd.DataFrame({
    'Metric': ['Accuracy', 'ROC-AUC', 'F1-Score'],
    'Train Score': [
        accuracy_score(y_train_balanced, train_pred_tuned),
        roc_auc_score(y_train_balanced, train_proba_tuned),
        f1_score(y_train_balanced, train_pred_tuned)
    ],
    'Test Score': [
        accuracy_score(y_test, test_pred_tuned),
        roc_auc_score(y_test, test_proba_tuned),
        f1_score(y_test, test_pred_tuned)
    ]
})

overfit_check['Train Score'] = overfit_check['Train Score'].round(4)
overfit_check['Test Score'] = overfit_check['Test Score'].round(4)
overfit_check['Gap (Train - Test)'] = (overfit_check['Train Score'] - overfit_check['Test Score']).round(4)

print("=== Overfitting Evaluation ===")
display(overfit_check)

# 3. Quick verdict
acc_gap = overfit_check.loc[overfit_check['Metric'] == 'Accuracy', 'Gap (Train - Test)'].values[0]
if acc_gap < 0.10:
    print(f"\nVerdict: Model shows healthy generalization. Accuracy gap is {acc_gap:.4f} (under 10%).")
else:
    print(f"\nVerdict: Mild-to-moderate overfitting detected. Accuracy gap is {acc_gap:.4f}.")
```

```python
def predict_churn(customer_data, model=best_svm, scaler=ss, training_columns=new_data.columns):
  
    # 1. Convert input dictionary to DataFrame if needed
    if isinstance(customer_data, dict):
        df_cust = pd.DataFrame([customer_data])
    else:
        df_cust = customer_data.copy()

    # 2. Total Charges numeric conversion
    if 'Total Charges' in df_cust.columns:
        df_cust['Total Charges'] = pd.to_numeric(df_cust['Total Charges'], errors='coerce').fillna(0)

    # 3. Binary categorical mapping
    binary_maps = {
        'Gender': {'Male': 1, 'Female': 0},
        'Senior Citizen': {'Yes': 1, 'No': 0, 1: 1, 0: 0},
        'Partner': {'Yes': 1, 'No': 0},
        'Dependents': {'Yes': 1, 'No': 0},
        'Phone Service': {'Yes': 1, 'No': 0},
        'Paperless Billing': {'Yes': 1, 'No': 0}
    }
    for col, mapping in binary_maps.items():
        if col in df_cust.columns:
            df_cust[col] = df_cust[col].map(mapping)

    # 4. One-hot encoding for remaining categorical fields
    cat_cols = df_cust.select_dtypes(include=['object', 'string']).columns.tolist()
    if cat_cols:
        df_cust = pd.get_dummies(df_cust, columns=cat_cols, drop_first=True, dtype=int)

    # 5. Reindex to align exactly with the 30 features from training
    df_aligned = df_cust.reindex(columns=training_columns, fill_value=0)

    # 6. Apply the fitted StandardScaler
    scaled_input = scaler.transform(df_aligned)

    # 7. Predict class & probability
    pred_class = int(model.predict(scaled_input)[0])
    pred_proba = float(model.predict_proba(scaled_input)[0][1])

    # 8. Display formatted output
    status = "CHURN RISK" if pred_class == 1 else "RETAIN (No Churn)"
    print("=" * 45)
    print("       CUSTOMER CHURN PREDICTION RESULT      ")
    print("=" * 45)
    print(f"Prediction       : {status} (Class {pred_class})")
    print(f"Churn Probability: {pred_proba:.2%}")
    print(f"Confidence Level : {max(pred_proba, 1 - pred_proba):.2%}")
    print("=" * 45)

    return {
        "predicted_class": pred_class,
        "churn_probability": pred_proba
    }
```
